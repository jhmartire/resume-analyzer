```python
import os
import google.generativeai as genai
import fitz  # PyMuPDF para leitura de PDF

# Configurar chave da API
api_key = os.environ.get("GOOGLE_API_KEY")


# Configurar API Gemini
genai.configure(api_key=api_key)
model = genai.GenerativeModel('gemini-1.5-pro')  

def extract_text_from_pdf(pdf_path):
    """Extrai texto de um arquivo PDF."""
    try:
        doc = fitz.open(pdf_path)
        text = ""
        for page in doc:
            text += page.get_text("text") + "\n"
        return text.strip()
    except Exception as e:
        return f"Erro ao ler o PDF: {str(e)}"

def analyze_resume(pdf_path):
    """Lê e analisa um currículo em PDF, fornecendo sugestões de melhoria."""
    resume_text = extract_text_from_pdf(pdf_path)

    if not resume_text:
        return "Nenhum texto foi extraído do PDF. Verifique o arquivo."

    prompt = f"""
    Analise o seguinte currículo e forneça sugestões de melhoria para torná-lo mais atrativo para recrutadores.
    Considere aspectos como clareza, estrutura, experiência relevante e palavras-chave importantes para o setor de tecnologia.

    Texto do currículo:
    {resume_text}

    Dê sugestões de melhoria, identificando pontos fortes e fracos. 
    Se necessário, sugira reescrita de seções e inclusão de palavras-chave mais estratégicas.
    """

    response = model.generate_content(prompt)
    
    return response.text if response else "Erro ao gerar análise do currículo."

# Caminho correto do arquivo PDF
pdf_file = "/Users/jhmartire/repos/livecode/Profile.pdf"

# Verificar se o caminho está correto
print(f"Verificando o arquivo: {pdf_file}")

if os.path.exists(pdf_file):
    feedback = analyze_resume(pdf_file)
    print("\nSugestões para melhoria do currículo:\n")
    print(feedback)
else:
    print(f"Erro: O arquivo {pdf_file} não foi encontrado.")
```
