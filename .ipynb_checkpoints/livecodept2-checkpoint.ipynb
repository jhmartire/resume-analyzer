# IMPORT LIBRARY 


```python
import os
import google.generativeai as genai
import fitz  # PyMuPDF para leitura de PDF
from fpdf import FPDF  # Biblioteca para criar PDFs
```

# Gemini API Configuration

```python
# Configure the API key (set in the environment before running the code)
api_key = os.environ.get("GOOGLE_API_KEY")

# Gemini model configuration
genai.configure(api_key=api_key)
model = genai.GenerativeModel('gemini-1.5-pro')
```

# Function to Extract Text from PDF

```python
def extract_text_from_pdf(pdf_path):
    """Extract text from a PDF file."""
    try:
        doc = fitz.open(pdf_path)
        text = "\n".join([page.get_text("text") for page in doc])
        return text.strip()
    except Exception as e:
        return f"Erro ao ler o PDF: {str(e)}"
```

# Function to Analyze Resume with AI

```python
def analyze_resume(resume_text):
    """Analyzes the resume and provides suggestions for improvement."""
    if not resume_text:
        return "No text extracted from PDF. Check the file."

    prompt = f"""
    **Curriculum Analysis** 

    🔍 **Positive Points**  
    - Highlight the well-written parts.  
    - List the keywords that have already been used well.  

     **What can improve?**  
    - Suggest adjustments to the structure and organization.  
    - Please indicate industry keywords that can be added.  

    **Detailed suggestions**  
    - If necessary, suggest reformulations of sections.  
    - Give practical examples of improved sentences.  

    **CV text:**  
    {resume_text}
    """

    response = model.generate_content(prompt)
    return response.text if response else "Error generating resume analysis."
```

# Function to Generate a Professional PDF

```python
def generate_pdf(feedback, output_filename="Evaluated_Curriculum_Support.pdf"):
    """Generates a formatted PDF with improvement suggestions."""
    pdf = FPDF()
    pdf.set_auto_page_break(auto=True, margin=15)
    pdf.add_page()

    # Title setting
    pdf.set_font("Arial", "B", 16)
    pdf.cell(200, 10, "Curriculum Analysis Report", ln=True, align="C")

    pdf.ln(10)  # Space before content

    # Add formatted feedback
    pdf.set_font("Arial", size=12)
    
    for line in feedback.split("\n"):
        if "**" in line:  # If it's a title
            pdf.set_font("Arial", "B", 14)
            line = line.replace("**", "")  # Remove asterisks
        else:
            pdf.set_font("Arial", size=12)

        pdf.multi_cell(0, 10, line)
    
    output_path = f"/tmp/{output_filename}"
    pdf.output(output_path)
    return output_path
```

# Run and Test the Code

```python
# Set resume path (adjust as per your directory)
pdf_file = "/Users/jhmartire/repos/livecode/Profile.pdf"

# Check if file exists
if os.path.exists(pdf_file):
    print(" File found! Analyzing resume...\n")

    # Extract text
    resume_text = extract_text_from_pdf(pdf_file)

    # Analyze resume
    feedback = analyze_resume(resume_text)

    # Display result in notebook
    print("\n Improvement Suggestions:\n")
    print(feedback)

    # Generate the PDF with the suggestions
    pdf_path = generate_pdf(feedback)

    print(f"\n PDF successfully generated: {pdf_path}")
else:
    print(f" Erro: O file {pdf_file} was not found.")
```

```python
import os
import google.generativeai as genai
import fitz  # PyMuPDF para leitura de PDF

# Configurar chave da API
api_key = os.environ.get("GOOGLE_API_KEY")


# Configurar API Gemini
genai.configure(api_key=api_key)
model = genai.GenerativeModel('gemini-1.5-pro')  

def extract_text_from_pdf(pdf_path):
    """Extrai texto de um arquivo PDF."""
    try:
        doc = fitz.open(pdf_path)
        text = ""
        for page in doc:
            text += page.get_text("text") + "\n"
        return text.strip()
    except Exception as e:
        return f"Erro ao ler o PDF: {str(e)}"

def analyze_resume(pdf_path):
    """Lê e analisa um currículo em PDF, fornecendo sugestões de melhoria."""
    resume_text = extract_text_from_pdf(pdf_path)

    if not resume_text:
        return "Nenhum texto foi extraído do PDF. Verifique o arquivo."

    prompt = f"""
    Analise o seguinte currículo e forneça sugestões de melhoria para torná-lo mais atrativo para recrutadores.
    Considere aspectos como clareza, estrutura, experiência relevante e palavras-chave importantes para o setor de tecnologia.

    Texto do currículo:
    {resume_text}

    Dê sugestões de melhoria, identificando pontos fortes e fracos. 
    Se necessário, sugira reescrita de seções e inclusão de palavras-chave mais estratégicas.
    """

    response = model.generate_content(prompt)
    
    return response.text if response else "Erro ao gerar análise do currículo."

# Caminho correto do arquivo PDF
pdf_file = "/Users/jhmartire/repos/livecode/Profile.pdf"

# Verificar se o caminho está correto
print(f"Verificando o arquivo: {pdf_file}")

if os.path.exists(pdf_file):
    feedback = analyze_resume(pdf_file)
    print("\nSugestões para melhoria do currículo:\n")
    print(feedback)
else:
    print(f"Erro: O arquivo {pdf_file} não foi encontrado.")
```
